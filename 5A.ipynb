```python
# a. Import libraries
import tensorflow as tf
import matplotlib.pyplot as plt
import numpy as np

from tensorflow.keras.applications import VGG16
from tensorflow.keras.applications.vgg16 import preprocess_input
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, GlobalAveragePooling2D
from tensorflow.keras.optimizers import Adam

# b. Load dataset
path = "caltech-101"

train = tf.keras.utils.image_dataset_from_directory(
    path,
    image_size=(224, 224),
    batch_size=32,
    validation_split=0.2,
    subset="training",
    seed=42
)

test = tf.keras.utils.image_dataset_from_directory(
    path,
    image_size=(224, 224),
    batch_size=32,
    validation_split=0.2,
    subset="validation",
    seed=42,
    shuffle=False
)

# Extract class names
classes = train.class_names
n = len(classes)

print("Classes:", classes)
print("Number of classes:", n)

# Keep original images for visualization
display_data = test.take(2)

# Reduce dataset samples
train = train.take(25)
test = test.take(2)

# Preprocess images for VGG16
train = train.map(lambda x, y: (preprocess_input(x), y))
test = test.map(lambda x, y: (preprocess_input(x), y))

# c. Load pretrained VGG16 model
base = VGG16(
    weights="imagenet",
    include_top=False,
    input_shape=(224, 224, 3)
)

# d. Freeze base layers
base.trainable = False

# Create Sequential model
model = tf.keras.models.Sequential([
    base,
    GlobalAveragePooling2D(),
    Dense(128, activation="relu"),
    Dense(n, activation="softmax")
])

model.compile(
    optimizer=Adam(0.001),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()

# e. Train model
history = model.fit(
    train,
    validation_data=test,
    epochs=5
)

# f. Compute loss
print("Training Loss:", history.history["loss"][-1])
print("Validation Loss:", history.history["val_loss"][-1])

# g. Evaluate model
loss, accuracy = model.evaluate(test)

print("Test Loss:", loss)
print("Test Accuracy:", accuracy)

# h. Plot accuracy and loss

plt.plot(history.history["accuracy"], label="Training Accuracy")
plt.plot(history.history["val_accuracy"], label="Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.show()

plt.plot(history.history["loss"], label="Training Loss")
plt.plot(history.history["val_loss"], label="Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.show()

# Visualize one prediction using original images
for images, labels in display_data.take(1):

    prediction = model.predict(
        preprocess_input(images),
        verbose=0
    )

    i = 0
    predicted_class = classes[np.argmax(prediction[i])]

    plt.imshow(images[i].numpy().astype("uint8"))
    plt.title("Predicted: " + predicted_class)
    plt.axis("off")
    plt.show()

# i. Fine-tune the model
base.trainable = True

for layer in base.layers[:-4]:
    layer.trainable = False

model.compile(
    optimizer=Adam(0.00001),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

history2 = model.fit(
    train,
    validation_data=test,
    epochs=3
)

# Final evaluation
loss, accuracy = model.evaluate(test)

print("Fine-tuned Loss:", loss)
print("Fine-tuned Accuracy:", accuracy)
```
