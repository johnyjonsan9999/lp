```python
import pandas as pd
import re
from gensim.models import Word2Vec

df=pd.read_csv("twitter_dataset.csv")

tweets = df['Text'].dropna()
sentences=[]

for tweet in tweets:
    tweet = re.sub(r"[^a-zA-Z\s]", "", str(tweet))
    words = tweet.lower().split()

    if words:
        sentences.append(words)

w2v = Word2Vec(
    sentences,
    vector_size=50,
    window=2,
    min_count=2,
    sg=0,
    epochs=5
)

word="good"


if word in w2v.wv:
    print("Word Embedding:",w2v.wv[word])
    print("Similar Words:",w2v.wv.most_similar(word,topn=5))

else:
    print("Word not found in vocabulary")

print("V Size:",len(w2v.wv))
```
