```python
import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Rescaling, Dense, Flatten
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.applications import VGG16
from tensorflow.keras.utils import image_dataset_from_directory
```

```python
train_data = image_dataset_from_directory(
    r"C:\Users\Debadwar\Desktop\LP-3 Deep Learning\seg_train\seg_train",
    image_size=(224, 224),
    batch_size = 32
)

test_data = image_dataset_from_directory(
    r"C:\Users\Debadwar\Desktop\LP-3 Deep Learning\seg_test\seg_test",
    image_size=(224, 224),
    batch_size=32
)
```

```python
train_data = train_data.take(30)
test_data = test_data.take(6)
```

```python
base_model = VGG16(
    weights = "imagenet",
    include_top=False,
    input_shape=(224, 224, 3)
)
```

```python
base_model.trainable = False
```

```python
x = Rescaling(1./255)(base_model.input)
x = base_model(x)

x = Flatten()(x)
x = Dense(128, activation = "relu")(x)
output = Dense(6, activation = "softmax")(x)

model = Model(base_model.input, output)
```

```python
model.compile(
    optimizer = Adam(learning_rate = 0.001),
    loss = "sparse_categorical_crossentropy",
    metrics = ["accuracy"]
)
```

```python
history = model.fit(
    train_data, 
    validation_data = test_data,
    epochs = 6
)
```

```python
loss, accuracy = model.evaluate(test_data)
print("loss: ", loss)
print("accuracy: ", accuracy)
```

```python
base_model.trainable = True
```
